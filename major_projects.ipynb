```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from google.colab import files

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")

display(df.head())
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
display(df.dtypes)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe(include="all").T)

print("\nMissing Values:")
missing_values = df.isnull().sum()

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage
})

display(missing_summary[missing_summary["Missing Values"] > 0])

print("\nDuplicate Records:")
print(df.duplicated().sum())

print("\nUnique Values:")
unique_summary = pd.DataFrame({
    "Column": df.columns,
    "Unique Values": [df[col].nunique() for col in df.columns]
})

display(unique_summary)
```

```python
missing = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df)) * 100
})

print("MISSING VALUES")
display(missing)

print("\nDUPLICATE RECORDS")
print("Number of duplicate rows:", df.duplicated().sum())

print("\nUNIQUE VALUES")
unique_values = pd.DataFrame({
    "Column": df.columns,
    "Unique Values": [df[col].nunique() for col in df.columns]
})

display(unique_values)
```

```python
data = df.copy()

duplicates_before = data.duplicated().sum()
data = data.drop_duplicates()

print("Duplicate rows removed:", duplicates_before)

data.columns = data.columns.str.strip()

categorical_columns = data.select_dtypes(include="object").columns

for col in categorical_columns:
    data[col] = data[col].astype(str).str.strip()

print("\nMissing values after cleaning:")
display(data.isnull().sum())

print("\nCleaned dataset shape:", data.shape)
```

```python
numerical_columns = data.select_dtypes(
    include=np.number
).columns.tolist()

categorical_columns = data.select_dtypes(
    include="object"
).columns.tolist()

print("NUMERICAL VARIABLES")
print("-" * 40)

for col in numerical_columns:
    print(col)

print("\nCATEGORICAL VARIABLES")
print("-" * 40)

for col in categorical_columns:
    print(col)

print("\nNumber of numerical columns:", len(numerical_columns))
print("Number of categorical columns:", len(categorical_columns))
```

```python
for col in categorical_columns:
    print("\n" + "=" * 60)
    print("COLUMN:", col)
    print("Number of unique values:", data[col].nunique())
    print("\nValue counts:")
    display(data[col].value_counts().head(20))
```

```python
print("SEASON DISTRIBUTION")

season_counts = data["Season"].value_counts()

display(season_counts.to_frame("Number of Records"))

plt.figure(figsize=(8, 5))

plt.bar(
    season_counts.index,
    season_counts.values
)

plt.title("Distribution of Agricultural Records by Season")
plt.xlabel("Season")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
season_summary = data.groupby("Season").agg(
    Average_Yield_Tonnes_Ha=("Yield_Tonnes_Ha", "mean"),
    Average_Cost_INR=("Total_Cost_INR", "mean"),
    Average_Revenue_INR=("Revenue_INR", "mean"),
    Average_Profit_INR=("Profit_INR", "mean"),
    Average_Water_Used_m3=("Water_Used_m3", "mean")
).round(2)

display(season_summary)
```

```python
yield_by_season = data.groupby("Season")["Yield_Tonnes_Ha"].mean().sort_values(ascending=False)

print("Average Yield by Season:")
display(yield_by_season.to_frame("Average Yield"))

plt.figure(figsize=(9, 5))

plt.bar(
    yield_by_season.index,
    yield_by_season.values
)

plt.title("Average Agricultural Yield Across Seasons")
plt.xlabel("Season")
plt.ylabel("Average Yield")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
profit_by_season = data.groupby("Season")["Profit_INR"].mean().sort_values(
    ascending=False
)

print("Average Profit by Season:")
display(profit_by_season.to_frame("Average Profit"))

plt.figure(figsize=(9, 5))

plt.bar(
    profit_by_season.index,
    profit_by_season.values
)

plt.title("Average Agricultural Profit Across Seasons")
plt.xlabel("Season")
plt.ylabel("Average Profit")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
economic_analysis = data.groupby("Season").agg(
    Average_Cost=("Total_Cost_INR", "mean"),
    Average_Revenue=("Revenue_INR", "mean"),
    Average_Profit=("Profit_INR", "mean")
).round(2)

display(economic_analysis)

economic_analysis.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Economic Performance Across Seasons")
plt.xlabel("Season")
plt.ylabel("Average Value")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
data["Profit_Margin"] = np.where(
    data["Revenue_INR"] != 0,
    (data["Profit_INR"] / data["Revenue_INR"]) * 100,
    0
)

profit_margin_by_season = data.groupby(
    "Season"
)["Profit_Margin"].mean().sort_values(ascending=False)

print("Average Profit Margin by Season:")
display(
    profit_margin_by_season.to_frame("Average Profit Margin (%)").round(2)
)

plt.figure(figsize=(9, 5))

plt.bar(
    profit_margin_by_season.index,
    profit_margin_by_season.values
)

plt.title("Average Profit Margin Across Seasons")
plt.xlabel("Season")
plt.ylabel("Profit Margin (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
resource_columns = [
    col for col in [
        "Water_Used_m3",
        "Fertilizer_kg_ha",
        "Pesticide_Litre_ha"
    ]
    if col in data.columns
]

if resource_columns:

    resource_by_season = data.groupby(
        "Season"
)[resource_columns].mean().round(2)

    print("Average Resource Usage by Season:")
    display(resource_by_season)

    resource_by_season.plot(
        kind="bar",
        figsize=(12, 6)
    )

    plt.title("Resource Usage Across Seasons")
    plt.xlabel("Season")
    plt.ylabel("Average Usage")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
else:
    print("Resource columns were not found.")
```

```python
environmental_columns = [
    col for col in [
        "Avg_Temperature_C",
        "Rainfall_mm",
        "Humidity_pct",
        "Soil_Moisture_pct"
    ]
    if col in data.columns
]

if environmental_columns:

    environmental_by_season = data.groupby(
        "Season"
)[environmental_columns].mean().round(2)

    print("Environmental Conditions by Season:")
    display(environmental_by_season)

    environmental_by_season.plot(
        kind="bar",
        figsize=(12, 6)
    )

    plt.title("Environmental Conditions Across Seasons")
    plt.xlabel("Season")
    plt.ylabel("Average Value")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
else:
    print("Environmental columns were not found.")
```

```python
for col in environmental_columns:

    plt.figure(figsize=(8, 5))

    plt.scatter(
        data[col],
        data["Yield_Tonnes_Ha"],
        alpha=0.5
    )

    plt.title(f"{col} vs Agricultural Yield")
    plt.xlabel(col)
    plt.ylabel("Yield")
    plt.tight_layout()
    plt.show()
```

```python
if "Water_Used_m3" in data.columns:

    plt.figure(figsize=(8, 5))

    plt.scatter(
        data["Water_Used_m3"],
        data["Yield_Tonnes_Ha"],
        alpha=0.5
    )

    plt.title("Water Usage vs Agricultural Yield")
    plt.xlabel("Water Usage")
    plt.ylabel("Yield")
    plt.tight_layout()
    plt.show()
```

```python
if "Crop" in data.columns:

    crop_season_yield = pd.pivot_table(
        data,
        values="Yield_Tonnes_Ha",
        index="Crop",
        columns="Season",
        aggfunc="mean"
    )

    print("Average Yield by Crop and Season:")
    display(crop_season_yield.round(2))

    crop_season_yield.plot(
        kind="bar",
        figsize=(14, 7)
    )

    plt.title("Crop-wise Yield Across Seasons")
    plt.xlabel("Crop")
    plt.ylabel("Average Yield")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
region_column = None

for col in ["State", "District", "Region", "Location", "Area", "Geographical_Area"]:
    if col in data.columns:
        region_column = col
        break

if region_column:

    region_season = pd.pivot_table(
        data,
        values="Yield_Tonnes_Ha",
        index=region_column,
        columns="Season",
        aggfunc="mean"
    )

    print(f"Average Yield by {region_column} and Season:")
    display(region_season.round(2))

    region_season.plot(
        kind="bar",
        figsize=(14, 7)
    )

    plt.title(f"Yield Across Seasons by {region_column}")
    plt.xlabel(region_column)
    plt.ylabel("Average Yield")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
else:
    print("No region/location column found.")
```

```python
def find_outliers(column):

    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower) |
        (data[column] > upper)
    ]

    return outliers, lower, upper


for col in numerical_columns:

    outliers, lower, upper = find_outliers(col)

    print(
        f"{col}: {len(outliers)} outliers "
        f"({len(outliers)/len(data)*100:.2f}%)"
    )
```

```python
season_variability = data.groupby("Season").agg(
    Yield_Mean=("Yield_Tonnes_Ha", "mean"),
    Yield_Std=("Yield_Tonnes_Ha", "std"),
    Profit_Mean=("Profit_INR", "mean"),
    Profit_Std=("Profit_INR", "std")
).round(2)

display(season_variability)
```

```python
seasons = data["Season"].dropna().unique()

yield_values = [
    data.loc[data["Season"] == season, "Yield_Tonnes_Ha"].dropna()
    for season in seasons
]

plt.figure(figsize=(10, 6))

plt.boxplot(
    yield_values,
    labels=seasons
)

plt.title("Yield Distribution Across Seasons")
plt.xlabel("Season")
plt.ylabel("Yield")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
seasons = data["Season"].dropna().unique()

profit_values = [
    data.loc[data["Season"] == season, "Profit_INR"].dropna()
    for season in seasons
]

plt.figure(figsize=(10, 6))

plt.boxplot(
    profit_values,
    labels=seasons
)

plt.title("Profit Distribution Across Seasons")
plt.xlabel("Season")
plt.ylabel("Profit")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
from scipy.stats import f_oneway

season_groups = [
    group["Yield_Tonnes_Ha"].dropna()
    for name, group in data.groupby("Season")
]

anova_stat, p_value = f_oneway(*season_groups)

print("One-Way ANOVA")
print("=" * 40)

print("F-statistic:", round(anova_stat, 4))
print("P-value    :", round(p_value, 6))

if p_value < 0.05:
    print("\nResult: Significant difference exists between")
    print("average yield across seasons.")
else:
    print("\nResult: No statistically significant difference")
    print("was detected between seasons.")
```

```python
profit_groups = [
    group["Profit_INR"].dropna()
    for name, group in data.groupby("Season")
]

anova_stat_profit, p_value_profit = f_oneway(
    *profit_groups
)

print("One-Way ANOVA — Profit")
print("=" * 40)

print("F-statistic:", round(anova_stat_profit, 4))
print("P-value    :", round(p_value_profit, 6))

if p_value_profit < 0.05:
    print("\nResult: Significant difference exists between")
    print("average profit across seasons.")
else:
    print("\nResult: No statistically significant difference")
    print("was detected between seasons.")
```

```python
final_summary = data.groupby("Season").agg(
    Records=("Season", "count"),
    Avg_Yield=("Yield_Tonnes_Ha", "mean"),
    Avg_Cost=("Total_Cost_INR", "mean"),
    Avg_Revenue=("Revenue_INR", "mean"),
    Avg_Profit=("Profit_INR", "mean"),
    Avg_Water_Usage=("Water_Used_m3", "mean"),
    Avg_Profit_Margin=("Profit_Margin", "mean")
).round(2)

display(final_summary)
```

```python
best_yield = final_summary["Avg_Yield"].idxmax()
worst_yield = final_summary["Avg_Yield"].idxmin()

best_profit = final_summary["Avg_Profit"].idxmax()
worst_profit = final_summary["Avg_Profit"].idxmin()

best_margin = final_summary["Avg_Profit_Margin"].idxmax()

print("=" * 70)
print("KEY FINDINGS — SEASONAL AGRICULTURE PERFORMANCE")
print("=" * 70)

print(f"\n1. Highest average yield:")
print(f"   {best_yield} ({final_summary.loc[best_yield, 'Avg_Yield']:.2f})")

print(f"\n2. Lowest average yield:")
print(f"   {worst_yield} ({final_summary.loc[worst_yield, 'Avg_Yield']:.2f})")

print(f"\n3. Highest average profit:")
print(f"   {best_profit} ({final_summary.loc[best_profit, 'Avg_Profit']:.2f})")

print(f"\n4. Lowest average profit:")
print(f"   {worst_profit} ({final_summary.loc[worst_profit, 'Avg_Profit']:.2f})")

print(f"\n5. Highest average profit margin:")
print(f"   {best_margin} ({final_summary.loc[best_margin, 'Avg_Profit_Margin']:.2f}%)")

print("\n" + "=" * 70)
```
